```python
import pandas as pd
from itertools import product
data = []

for P, Q, R in product([True, False], repeat=3):

    S1 = (not Q) or P
    S2 = (not P) or (not Q)
    S3 = Q or R


    KB = S1 and S2 and S3

    S4 = (not R) or P
    S5 = (not Q) or R

    data.append([P, Q, R, S1, S2, S3, KB, S4, S5])

columns = ['P', 'Q', 'R', 'Q -> P', 'P -> ~Q',
           'Q OR R', 'KB', 'R -> P', 'Q -> R']

df = pd.DataFrame(data, columns=columns)

df = df.map(lambda x: 'T' if x else 'F')

print("TRUTH TABLE")
display(df)

models = df[df['KB'] == 'T']

print("\nMODELS OF KB")
display(models[['P', 'Q', 'R']])


print("\nENTAILMENT CHECKING")

print("ii) KB entails R:",
      all(models['R'] == 'T'))

print("iii) KB entails R -> P:",
      all(models['R -> P'] == 'T'))

print("iv) KB entails Q -> R:",
      all(models['Q -> R'] == 'T'))
print("Arbiya Nurain")
print("1WN24CS048")
```
