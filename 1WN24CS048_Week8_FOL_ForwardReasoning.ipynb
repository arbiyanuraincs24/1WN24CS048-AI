```python

facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

rules = [
    (["Enemy"], "Hostile"),
    (["Missile"], "Weapon"),
    (["Missile", "Owns"], "Sells"),
]

def forward_chaining(facts):
    changed = True

    print("\nInitial Facts:")
    for fact in facts:
        print(fact)

    print("\nForward Chaining Process:\n")

    while changed:
        changed = False

        for fact in list(facts):

            if fact[0] == "Enemy":
                new_fact = ("Hostile", fact[1])

                if new_fact not in facts:
                    facts.add(new_fact)
                    print("Enemy(A, America)")
                    print("       |")
                    print("       v")
                    print("Hostile(A)\n")
                    changed = True

            if fact[0] == "Missile":
                new_fact = ("Weapon", fact[1])

                if new_fact not in facts:
                    facts.add(new_fact)
                    print("Missile(T1)")
                    print("     |")
                    print("     v")
                    print("Weapon(T1)\n")
                    changed = True

        if ("Missile", "T1") in facts and ("Owns", "A", "T1") in facts:
            new_fact = ("Sells", "Robert", "T1", "A")

            if new_fact not in facts:
                facts.add(new_fact)
                print("Missile(T1) AND Owns(A,T1)")
                print("          |")
                print("          v")
                print("Sells(Robert,T1,A)\n")
                changed = True

        if (("American", "Robert") in facts and
            ("Weapon", "T1") in facts and
            ("Sells", "Robert", "T1", "A") in facts and
            ("Hostile", "A") in facts):

            new_fact = ("Criminal", "Robert")

            if new_fact not in facts:
                facts.add(new_fact)

                print("American(Robert) AND")
                print("Weapon(T1) AND")
                print("Sells(Robert,T1,A) AND")
                print("Hostile(A)")
                print("       |")
                print("       v")
                print("Criminal(Robert)\n")

                changed = True

    return facts


result = forward_chaining(facts)

if ("Criminal", "Robert") in result:
    print("================================")
    print("       GOAL PROVED")
    print("================================")
    print("Criminal(Robert)")
else:
    print("Goal cannot be proved")
print("Arbiya Nurain")
print("1WN24CS048")
```
