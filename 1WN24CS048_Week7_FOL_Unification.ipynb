```python

def is_variable(x):
    return isinstance(x, str) and x[0].islower()

def occurs(var, term, subs):
    term = substitute(term, subs)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, t, subs) for t in term[1])

    return False

def substitute(term, subs):
    if isinstance(term, str):
        while term in subs:
            term = subs[term]
        return term

    if isinstance(term, tuple):
        return (term[0], tuple(substitute(t, subs) for t in term[1]))

    return term

def unify(x, y, subs):
    x = substitute(x, subs)
    y = substitute(y, subs)

    if x == y:
        return subs

    if is_variable(x):
        if occurs(x, y, subs):
            return None
        subs[x] = y
        return subs

    if is_variable(y):
        if occurs(y, x, subs):
            return None
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return None

        for a, b in zip(x[1], y[1]):
            subs = unify(a, b, subs)
            if subs is None:
                return None

        return subs

    return None

def parse_term(s):
    s = s.strip()

    if "(" not in s:
        return s

    name, rest = s.split("(", 1)
    rest = rest[:-1]

    args = []
    depth = 0
    start = 0

    for i, ch in enumerate(rest):
        if ch == "(":
            depth += 1
        elif ch == ")":
            depth -= 1
        elif ch == "," and depth == 0:
            args.append(parse_term(rest[start:i]))
            start = i + 1

    args.append(parse_term(rest[start:]))

    return (name.strip(), tuple(args))

def parse_expression(s):
    name, rest = s.split("(", 1)
    rest = rest[:-1]

    args = []
    depth = 0
    start = 0

    for i, ch in enumerate(rest):
        if ch == "(":
            depth += 1
        elif ch == ")":
            depth -= 1
        elif ch == "," and depth == 0:
            args.append(parse_term(rest[start:i]))
            start = i + 1

    args.append(parse_term(rest[start:]))

    return (name.strip(), tuple(args))

def display(term):
    if isinstance(term, tuple):
        return term[0] + "(" + ", ".join(display(x) for x in term[1]) + ")"
    return term

def unification(e1, e2):
    p1 = parse_expression(e1)
    p2 = parse_expression(e2)

    if p1[0] != p2[0] or len(p1[1]) != len(p2[1]):
        print("Unification Failed")
        return

    subs = {}

    for a, b in zip(p1[1], p2[1]):
        subs = unify(a, b, subs)

        if subs is None:
            print("Unification Failed")
            return

    print("Unification Successful")
    print("Substitutions:")

    for key, value in subs.items():
        print(key, "=", display(substitute(value, subs)))

    result = (p1[0], tuple(substitute(x, subs) for x in p1[1]))
    print("Unified Expression:", display(result))


e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

unification(e1, e2)
print("Arbiya Nurain")
print("1WN24CS048")
```
