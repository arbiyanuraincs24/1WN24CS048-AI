```python

def calculate_cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def hill_climbing_4_queens(initial_state):
    current_state = list(initial_state)
    current_cost = calculate_cost(current_state)

    step = 1

    # Modified tie-breaking order to obtain (2, 0, 3, 1)
    pairs = [(0, 2), (0, 1), (0, 3),
             (1, 2), (1, 3), (2, 3)]

    while True:
        print("\n--- STEP", step, "---")
        print("Current State:", current_state)
        print("Current Cost:", current_cost)

        best_neighbor = None
        best_cost = current_cost
        best_pair = None

        print("\nNeighbour Costs:")

        for i, j in pairs:
            neighbor = current_state.copy()

            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

            neighbor_cost = calculate_cost(neighbor)

            print("Swap (x_%d, x_%d) -> %s | Cost: %d"
                  % (i, j, neighbor, neighbor_cost))

            if neighbor_cost < best_cost:
                best_cost = neighbor_cost
                best_neighbor = neighbor
                best_pair = (i, j)

        if best_neighbor is None:
            print("\nAlgorithm Terminated")
            print("Final State:", current_state)
            print("Final Cost:", current_cost)

            if current_cost == 0:
                print("Goal Configuration Reached!")
            else:
                print("Stuck in Local Optimum")

            break

        print("\nMoving to best neighbour by swapping",
              best_pair)

        current_state = best_neighbor
        current_cost = best_cost

        step += 1


# Initial State
initial_state = [3, 1, 2, 0]

hill_climbing_4_queens(initial_state)
print("Arbiya Nurain")
print("1WN24CS048")
```
